Tensor Learning

```python
import torch

# 0阶张量
t0 = torch.tensor(0.45)
print(f"0阶: {t0.shape}")              
print(t0)
print("")

# 1阶张量 (向量) - 词嵌入向量
t1 = torch.randn(1)                    # [768] - 单个token的embedding
print(f"1阶: {t1.shape}")               
print(t1)
print("")

# 2阶张量 (矩阵) - 批次词嵌入 / 权重矩阵
t2 = torch.randn(3, 2)                # [batch=32, hidden=768]
print(f"2阶: {t2.shape}")           
print(t2)
print("")

t2b = torch.randn(1, 2)                # [batch=32, hidden=768]
print(f"t2b 2阶: {t2b.shape}")           
print(t2b)
print("")

# 3阶张量 - 序列数据 / Attention输出
t3 = torch.randn(3, 3, 2)           # [batch=32, seq_len=128, hidden=768]
print(f"3阶: {t3.shape}")          
print(t3)
print("")

# 4阶张量 - 图像批次 / 多头Attention
t4 = torch.randn(4, 3, 5, 2)         # [batch=32, heads=12, seq_len=64, head_dim=64]
print(f"4阶: {t4.shape}")             
print(t4)
print("")

# 5阶张量 - 视频数据 / 3D卷积输入
t5 = torch.randn(3, 3, 5, 7, 3)    # [batch=16, channels=3, frames=16, H=224, W=224]
print(f"5阶: {t5.shape}")        
print(t5)
```
